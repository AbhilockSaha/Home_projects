# Linear Algebra Practice

```python
a = [2,4,6]
b = [1,3,5]
x = []
i=0
while(i < (len(a)) ):
    x.append(a[i] + b[i])
    i+=1

print(x)
    
      
```

# Substraction

```python
a = [10,8,6]
b = [2,3,4]
x = []
i=0
while(i < (len(a)) ):
    x.append(a[i] - b[i])
    i+=1

print(x)
    
```

# Matrix Multiplication

```python
a = [2, 4, 6]
k = 3
x = []
i=0
while(i < (len(a)) ):
    x.append(a[i]*k)
    i+=1
  
print(x)
```

# Vector Magnitude

```python
a = [3, 4]
i = 0
s = 0

while i < len(a):
    s += a[i]**2
    i += 1

print(s**(1/2))
```

# Dot Product 

```python
A = [1,2,3]
B = [4,5,6]
dot = 0
for i in range (len(A)):
    mul = A[i] * B[i]
    dot += mul
print("Dot product : ",dot)
```

# Orthogonality 

```python
A = [1,2]
B = [2,-1]
dot = 0
for i in range (len(A)):
    mul = A[i] * B[i]
    dot += mul
if dot :
    print("Not orthogonal")
else:
    print("Orthogonal")
```

# Matrix Addition

```python
A = [
    [1, 2],
    [3, 4]
]

B = [
    [5, 6],
    [7, 8]
]
k = []
for i in range(len(A)):
    row =[]
    for j in range(len(A)):
        row.append(A[i][j] + B[i][j])

    k.append(row)
        
print(k)
```

# MAtrix Substraction

```python
A = [
    [1, 2],
    [3, 4]
]

B = [
    [5, 6],
    [7, 8]
]
k = []
for i in range(len(A)):
    row =[]
    for j in range(len(A)):
        row.append(A[i][j] - B[i][j])

    k.append(row)
        
print(k)
```

# Matrix Scalar Multiplication

```python
A = [
    [1, 2 , 5],
    [3, 4 , 6]
]
x=[]
k = 5
for i in range(len(A)):
    row =[]
    for j in range(len(A[i])):
        row.append(A[i][j] * k)
    x.append(row)

print(x)
```
